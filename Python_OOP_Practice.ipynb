```python
class Car:
    def __init__(self, brand: str, model: str, fuel: float, consumption_rate: float = 0.08):
        self.brand = brand
        self.model = model
        self.fuel = fuel  # in liters
        self.consumption_rate = consumption_rate  # liters consumed per km

    def drive(self, distance: float):
        fuel_needed = distance * self.consumption_rate

        if self.fuel >= fuel_needed:
            self.fuel -= fuel_needed
            print(f"Drove {distance} km. Remaining fuel: {self.fuel:.2f}L")
        else:
            max_distance = self.fuel / self.consumption_rate
            self.fuel = 0.0
            print(f"Not enough fuel! Drove only {max_distance:.2f} km before running out.")

    def refuel(self, amount: float):
        self.fuel += amount
        print(f"Added {amount}L of fuel. Current fuel: {self.fuel:.2f}L")
```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.available = True  # True if available, False if borrowed

class Library:
    def __init__(self):
        self.books = []  # List to store Book objects

    def add_book(self, title, author):
        new_book = Book(title, author)
        self.books.append(new_book)
        print(f"Book '{title}' added successfully.")

    def borrow_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if book.available:
                    book.available = False
                    print(f"You have borrowed '{title}'.")
                    return
                else:
                    print(f"Sorry, '{title}' is already borrowed.")
                    return
        print(f"Book '{title}' not found in the library.")

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if not book.available:
                    book.available = True
                    print(f"Thank you for returning '{title}'.")
                    return
                else:
                    print(f"'{title}' was not checked out.")
                    return
        print(f"Book '{title}' not found in the library.")
```

```python
class Player:
    def __init__(self, name, runs):
        self.name = name
        self.runs = runs

class Team:
    def __init__(self):
        self.players = []

    def add_player(self, player):
        self.players.append(player)

    def highest_score(self):
        # 1. Guard against empty lists
        if not self.players:
            print("No players in the team.")
            return None

        # 2. Track the player with the highest runs
        highest = self.players[0]
        for player in self.players:
            if player.runs > highest.runs: # Fixed typo here (highesAt -> highest)
                highest = player

        return highest.name
```

```python
import math

# 1. Base Class (The Blueprint)
class Shape:
    def get_area(self):
        # Every shape must implement this, or it raises an error
        raise NotImplementedError("Subclasses must implement this method")

# 2. Child Classes (Inherit from Shape)
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def get_area(self):
        return math.pi * (self.radius ** 2)

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def get_area(self):
        return self.length * self.width

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def get_area(self):
        return 0.5 * self.base * self.height
```

```python
class Account:
    def __init__(self, balance):
        self.balance = balance

    def withdrawal(self, amount, withdrawal_limit):
        while amount > 0:
            if amount <= withdrawal_limit:
                if amount <= self.balance:
                    print(f"The amount {amount} is withdrawn")
                    self.balance -= amount
                    break  # Added to prevent an infinite loop
                else:
                    print("Insufficient funds for this withdrawal")
                    break  # Added to exit the loop
            else:
                print("The amount to withdraw is exceeding the limits")
                break  # Added to exit the loop

class Savings(Account):
    def __init__(self, amount, withdrawal_limit):
        super().__init__(amount)
        self.withdrawal_limit = withdrawal_limit

class Current(Account):
    def __init__(self, amount, withdrawal_limit):
        super().__init__(amount)
        self.withdrawal_limit = withdrawal_limit
```

```python
def is_prime(n):
    if n <= 1:
        return False
    # Check divisibility up to the square root of n
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Example usage:
print(is_prime(11))  # Output: True
print(is_prime(4))   # Output: False
```

```python
from abc import ABC,abstractmethod
class Paymentsystem(ABC):
  @abstractmethod
  def pay(self):
    pass
class PayPal(Paymentsystem):
  def pay(self,amount):
    return f"pay this {amount} Through paypal"
class GooglePay(Paymentsystem):
  def pay(self,amount):
     return f"pay this {amount} Through Googlepay"
class Phonepay(Paymentsystem):
  def pay(self,amount):
     return f"pay this {amount} Through Phonepay"
```
